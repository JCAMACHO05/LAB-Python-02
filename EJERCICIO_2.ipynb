```python
# Lista de ejemplo
numeros = [5, 10, 15, 20, 25]
temperaturas_c = [0, 20, 30, 40]
palabras = ["Python", "lista", "computador", "sol", "programacion"]
matriz = [[1,2,3], [4,5,6], [7,8,9]]

# a. Generar una lista con tuplas (n, n^2)
lista_tuplas = [(n, n**2) for n in numeros]
print("a.", lista_tuplas)

# b. Generar una lista con cubos de los números mayores a 10
cubos = [n**3 for n in numeros if n > 10]
print("b.", cubos)

# c. Generar una lista con solo los números pares
pares = [n for n in numeros if n % 2 == 0]
print("c.", pares)

# d. Generar lista tipo ["valor: 10", "valor: 15", ...]
valores = [f"valor: {n}" for n in numeros]
print("d.", valores)

# e. Conversión de Celsius a Fahrenheit
fahrenheit = [(c * 9/5) + 32 for c in temperaturas_c]
print("e.", fahrenheit)

# f. Filtrar palabras de más de 5 letras
largas = [p for p in palabras if len(p) > 5]
print("f.", largas)

# g. Obtener solo las iniciales de las palabras
iniciales = [p[0] for p in palabras]
print("g.", iniciales)

# h. Generar lista indicando "par" o "impar"
par_impar = ["par" if n % 2 == 0 else "impar" for n in numeros]
print("h.", par_impar)

# i. Aplanar una matriz
aplanada = [elemento for fila in matriz for elemento in fila]
print("i.", aplanada)
```
