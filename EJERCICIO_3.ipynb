```python
import math

# a. Convertir temperaturas de Celsius a Fahrenheit
celsius = [0, 20, 30, 40]

fahrenheit = [(c * 9/5) + 32 for c in celsius]

print("a.", fahrenheit)


# b. Sumar elementos de dos listas
lista1 = [1, 2, 3, 4]
lista2 = [10, 20, 30, 40]

suma_listas = [x + y for x, y in zip(lista1, lista2)]

print("b.", suma_listas)


# c. Generar lista con áreas de círculos
radios = [1, 2, 3, 4]

areas = [math.pi * r**2 for r in radios]

print("c.", areas)


# d. Generar lista de tuplas con aumento del 10%
productos = {
    "pan": 1000,
    "leche": 2500,
    "café": 5000
}

aumento = [(nombre, valor * 1.10) for nombre, valor in productos.items()]

print("d.", aumento)


# e. Multiplicar todos los elementos de una matriz por 10
matriz = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

matriz_x10 = [[elemento * 10 for elemento in fila] for fila in matriz]

print("e.", matriz_x10)
```
