```python
# Juego NIM usando Minimax
# MAX = Computadora
# MIN = Humano

import math

# Función minimax
def minimax(piedras, es_max):

    # Caso terminal
    if piedras == 0:
        if es_max:
            return -1   # MAX pierde
        else:
            return 1    # MIN pierde

    # Turno de MAX (Computadora)
    if es_max:
        mejor_valor = -math.inf

        for movimiento in [1, 2, 3]:
            if piedras >= movimiento:
                valor = minimax(piedras - movimiento, False)
                mejor_valor = max(mejor_valor, valor)

        return mejor_valor

    # Turno de MIN (Humano)
    else:
        mejor_valor = math.inf

        for movimiento in [1, 2, 3]:
            if piedras >= movimiento:
                valor = minimax(piedras - movimiento, True)
                mejor_valor = min(mejor_valor, valor)

        return mejor_valor


# Función para encontrar la mejor jugada de la CPU
def mejor_jugada(piedras):

    mejor_valor = -math.inf
    mejor_movimiento = 1

    for movimiento in [1, 2, 3]:

        if piedras >= movimiento:

            valor = minimax(piedras - movimiento, False)

            if valor > mejor_valor:
                mejor_valor = valor
                mejor_movimiento = movimiento

    return mejor_movimiento


# ----------------------------
# Juego principal
# ----------------------------

piedras = 10
turno_humano = True   # El humano empieza

print("=== JUEGO NIM ===")
print("Hay", piedras, "piedras.")
print("Puedes quitar 1, 2 o 3 piedras.\n")

while piedras > 0:

    print("Piedras restantes:", piedras)

    # Turno del humano
    if turno_humano:

        movimiento = int(input("¿Cuántas piedras deseas quitar? (1-3): "))

        # Validación
        while movimiento not in [1, 2, 3] or movimiento > piedras:
            movimiento = int(input("Movimiento inválido. Intenta de nuevo: "))

        piedras -= movimiento

        # Verificar derrota
        if piedras == 0:
            print("\nNo quedan piedras.")
            print("Perdiste :(")
            break

    # Turno de la CPU
    else:

        movimiento_cpu = mejor_jugada(piedras)

        print("La computadora quita", movimiento_cpu, "piedra(s).")

        piedras -= movimiento_cpu

        # Verificar derrota
        if piedras == 0:
            print("\nNo quedan piedras.")
            print("La computadora pierde.")
            print("¡Ganaste!")
            break

    # Cambiar turno
    turno_humano = not turno_humano
```
