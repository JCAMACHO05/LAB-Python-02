```python
# a. Filtrar números pares y luego elevarlos al cuadrado
numeros = [1, 2, 3, 4, 5, 6, 7, 8]

pares_cuadrado = sorted([n**2 for n in numeros if n % 2 == 0])

print("a.", pares_cuadrado)


# b. Elevar al cubo, filtrar mayores que 100 y convertir a cadenas
numeros2 = [2, 4, 5, 6, 7]

resultado = sorted([str(n**3) for n in numeros2 if n**3 > 100])

print("b.", resultado)


# c. Filtrar mayores que 10, elevar al cuadrado y ordenar de mayor a menor
numeros3 = [5, 12, 8, 15, 20, 3]

resultado2 = sorted(
    [n**2 for n in numeros3 if n > 10],
    reverse=True
)

print("c.", resultado2)
```
