```python
# a. Lambda que devuelve True si un número es múltiplo de 3
multiplo_de_3 = lambda x: x % 3 == 0

# b. Lambda que devuelve el cubo de un número
cubo = lambda x: x ** 3

# c. Lambda que devuelve el producto de dos números
producto = lambda x, y: x * y

# d. Lambda que devuelve el mayor de dos números
mayor = lambda x, y: x if x > y else y

# e. Lambda que devuelve True si una palabra empieza con "A" o "a"
empieza_con_a = lambda palabra: palabra.lower().startswith("a")

# f. Lambda que convierte Celsius a Fahrenheit
celsius_a_fahrenheit = lambda c: (c * 9/5) + 32


# Pruebas
print(multiplo_de_3(9))          # True
print(cubo(3))                   # 27
print(producto(4, 5))            # 20
print(mayor(10, 7))              # 10
print(empieza_con_a("Amigo"))    # True
print(celsius_a_fahrenheit(25))  # 77.0


# -----------------------------------------
# Ejercicio con lista de lambdas
# -----------------------------------------

# Lista con las tres lambdas
funciones = [
    lambda x: x * 2,     # Duplica el número
    lambda x: x + 10,    # Suma 10
    lambda x: x ** 2     # Calcula el cuadrado
]

# Pedir número al usuario
numero = int(input("Ingrese un número: "))

# Aplicar cada lambda y mostrar resultados
for funcion in funciones:
    print(funcion(numero))
```
