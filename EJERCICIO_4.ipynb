```python
# Listas de ejemplo
palabras = ["sol", "python", "casa", "computador", "oso", "ana", "radar"]
lista = [1, None, 5, None, 10, 20]
numeros = [15, 22, 35, 40, 55, 78]

# a. Filtrar palabras con más de 4 letras
mas_de_4 = [p for p in palabras if len(p) > 4]
print("a.", mas_de_4)

# b. Filtrar elementos que no sean None
sin_none = [x for x in lista if x is not None]
print("b.", sin_none)

# c. Filtrar palabras que empiezan por vocal
vocales = [p for p in palabras if p[0].lower() in "aeiou"]
print("c.", vocales)

# d. Filtrar palabras que son palíndromos
palindromos = [p for p in palabras if p.lower() == p.lower()[::-1]]
print("d.", palindromos)

# e. Filtrar números que terminan en 5
terminan_5 = [n for n in numeros if str(n).endswith("5")]
print("e.", terminan_5)

# f. Filtrar productos con precio mayor a 100
productos = [
    {"nombre": "Teclado", "precio": 80},
    {"nombre": "Mouse", "precio": 40},
    {"nombre": "Monitor", "precio": 300},
    {"nombre": "Webcam", "precio": 150}
]

mayores_100 = [p for p in productos if p["precio"] > 100]

print("f.", mayores_100)
```

```python
from functools import reduce

# a. Multiplicación de todos los números de una lista
numeros = [1, 2, 3, 4, 5]

multiplicacion = reduce(lambda x, y: x * y, numeros)

print("a.", multiplicacion)


# b. Concatenar palabras en una sola cadena
palabras = ["Hola", "mundo", "Python"]

concatenadas = reduce(lambda x, y: x + " " + y, palabras)

print("b.", concatenadas)


# c. Hallar el mayor número usando una función auxiliar
numeros2 = [10, 45, 7, 89, 23]

# Función auxiliar
def mayor(x, y):
    return x if x > y else y

mayor_numero = reduce(mayor, numeros2)

print("c.", mayor_numero)
```

```python
# a. Ordenar una lista de palabras alfabéticamente
palabras = ["python", "java", "casa", "árbol", "computador"]

ordenadas = sorted(palabras)

print("a.", ordenadas)


# b. Ordenar una lista de tuplas por el primer elemento
tuplas = [(3, "c"), (1, "a"), (2, "b")]

orden_tuplas = sorted(tuplas, key=lambda x: x[0])

print("b.", orden_tuplas)


# c. Ordenar productos por precio (mayor a menor) y luego por nombre
productos = [
    ("Camisa", 30),
    ("Pantalón", 55),
    ("Medias", 10),
    ("Chaqueta", 80)
]

orden_productos = sorted(productos, key=lambda x: (-x[1], x[0]))

print("c.", orden_productos)


# d. Ordenar palabras según su longitud
palabras2 = ["sol", "computador", "mesa", "python", "a"]

orden_longitud = sorted(palabras2, key=len)

print("d.", orden_longitud)


# e. Ordenar números: primero pares y luego impares
numeros = [7, 2, 9, 4, 1, 6, 3, 8]

orden_par_impar = sorted(numeros, key=lambda x: x % 2)

print("e.", orden_par_impar)


# f. Contar letras de una palabra y ordenar por frecuencia
palabra = "programacion"

# Crear diccionario de frecuencias
frecuencias = {letra: palabra.count(letra) for letra in palabra}

# Ordenar por frecuencia
orden_frecuencias = sorted(frecuencias.items(), key=lambda x: x[1], reverse=True)

print("f.", orden_frecuencias)


# g. Ordenar empleados por departamento y luego por edad
empleados = [
    ("Maria", "Ventas", 30),
    ("Luis", "Tecnología", 25),
    ("Ana", "Ventas", 25),
    ("Pedro", "Tecnología", 28)
]

orden_empleados = sorted(empleados, key=lambda x: (x[1], x[2]))

print("g.", orden_empleados)
```
